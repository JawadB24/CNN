```python
import numpy as np
from cnn.layers import Layer, Conv, BatchNorm, ReLU, Pooling, Flatten, Dense, Softmax
from cnn.neuralnet import NeuralNet
```

```python
NN = NeuralNet()
m = 2
n_c = 1
n_filters = 1
n_h = 5
n_w = 5
f_h = 2
f_w = 2
conv_stride = 1
padding = 1
p_h = 2
p_w = 2
pool_stride = 2
n_h_prime = ((n_h + 2 * padding - f_h) // conv_stride) + 1
n_w_prime = ((n_w + 2 * padding - f_w) // conv_stride) + 1
mp_h = ((n_h_prime - p_h) // pool_stride) + 1
mp_w = ((n_w_prime - p_w) // pool_stride) + 1

classes = 3

rng = np.random.default_rng(42) #same numbers generated each time
X = rng.normal(size=(m, n_c, n_h, n_w))

y = np.array([0, 2])

NN.layers = [Conv(n_filters, n_c, f_h, f_w, conv_stride, padding), BatchNorm(n_filters, 4), ReLU(), Pooling(p_h, p_w, pool_stride), Flatten(), Dense(n_filters * mp_h * mp_w, classes), Softmax()]
```

```python
NN.set_dtype(np.float64)
NN.gradient_checking(X, y)
NN.set_dtype(np.float32)
```

The error between the analytical and approximated gradients for each layer with learnable parameters is < 1e-7, a common threshold used to determine the accuracy of our backpropagation process
